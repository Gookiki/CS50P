```python
import re
import sys


def main():
    print(validate(input("IPv4 Address: ")))


def validate(ip):
    if re.search(r"^(([0-9]|[1-9][0-9]|1[0-9][0-9]|2[0-4][0-9]|25[0-5])\.){3}([0-9]|[1-9][0-9]|1[0-9][0-9]|2[0-4][0-9]|25[0-5])$", ip):
        return True
    else:
        return False



if __name__ == "__main__":
    main()
```

```python
import re
import sys


def main():
    stuff = (parse(input("HTML: ").strip()))
    print("https://youtu.be/" + stuff)


def parse(s):
    m = re.search(r'<iframe[^>]*src="(https?://(?:www\.)?youtube\.com/embed/([a-zA-Z0-9_-]+))"[^>]*></iframe>', s)
    if m:
        return m.group(2)
    else:
        return None



if __name__ == "__main__":
    main()
```

```python
import re
import sys


def main():
    print(convert(input("Hours: ")))


def convert(s):
    Time = re.compile(
        r"^([1-9]|1[0-2])"
        r"(?::([0-5][0-9]))?"
        r" (AM|PM)"
        r" to "
        r"([1-9]|1[0-2])"
        r"(?::([0-5][0-9]))?"
        r" (AM|PM)$"
    )
    maatch = Time.fullmatch(s)
    if not maatch:
        raise ValueError

    h1, m1, p1, h2, m2, p2 = maatch.groups()
    h1 = int(h1)
    h2 = int(h2)
    m1 = m1 or "00"
    m2 = m2 or "00"

    if p1 == "PM" and h1 != 12:
        h1 += 12
    if p1 == "AM" and h1 == 12:
        h1 = 0
    if p2 == "PM" and h2 != 12:
        h2 += 12
    if p2 == "AM" and h2 == 12:
        h2 = 0

    return f"{h1:02}:{m1} to {h2:02}:{m2}"


if __name__ == "__main__":
    main()
```

```python
word = input("What is your camel case word? ")

snakecase = str((""))

for letter in word:
    if letter.isupper():
        snakecase = snakecase + "_"
    snakecase = snakecase + letter.lower()

print(snakecase)
```

```python
price = 50
deposit = 0

while deposit < price:
    coin = int(input("Insert Coin, 5, 10, 25 only "))
    if coin == 5 or coin == 10 or coin == 25:
        deposit = deposit + coin
        if deposit < price:
            print(f"you still owe {price - deposit}" )
        else:
            continue
    else:
        print("That is not accepted")
        continue    
change_owed = deposit - price
print(f"change owed: {change_owed}")
```

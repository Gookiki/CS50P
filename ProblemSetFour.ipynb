```python
import random

target = random.randint(1, 100)

while True:
    guess = input("Make a guess: ")
    try:
        guess = int(guess)
        if target > guess:
            print("Too low")
        elif target < guess:
            print("Too high")
        elif target == guess:
            print("You're Right")
            break
    except:
        continue

    
```

```python
from datetime import date
import datetime as dt
import sys


def main(dateofbirth):
    try:
        birth = date.fromisoformat(dateofbirth)
        today = date.today()
        delta = today - birth 
        minutes = delta.total_seconds() / 60
        return(minutes)
    except ValueError:
        print("Invalid date")
        sys.exit(1)

    





if __name__ == "__main__":
    print(main(input("Date of Birth: ").strip()))
    print(main("2000-01-01"))
    print(main("1990-01-01"))
```

```python
import pytest

def test_minutes():
    assert main("2000-01-01") == 13891680.0
    assert main("1990-01-01") == 19150560.0

test_minutes()
```

```python
class Jar:
    def __init__(self, capacity=12):
        if not isinstance(capacity, int) or capacity < 0:
            raise ValueError("capacity must be a non-negative integer")
        self._capacity = capacity
        self._size = 0

    def __str__(self):
        return "🍪" * self._size

    def deposit(self, n):
        if not isinstance(n, int) or n < 0:
            raise ValueError("deposit amount must be a non-negative integer")
        if self._size + n > self._capacity:
            raise ValueError("cannot deposit more cookies than capacity")
        self._size += n

    def withdraw(self, n):
        if not isinstance(n, int) or n < 0:
            raise ValueError("withdraw amount must be a non-negative integer")
        if n > self._size:
            raise ValueError("not enough cookies to withdraw")
        self._size -= n

    @property
    def capacity(self):
        return self._capacity

    @property
    def size(self):
        return self._size
```

```python
# Notebook tests for Jar

jar = Jar(5)
assert jar.capacity == 5
assert jar.size == 0

jar.deposit(3)
assert jar.size == 3
assert str(jar) == "🍪🍪🍪"

jar.withdraw(2)
assert jar.size == 1
assert str(jar) == "🍪"

# invalid deposit
try:
    jar.deposit(10)
    raise AssertionError("deposit should have raised ValueError")
except ValueError:
    pass

# invalid withdraw
try:
    jar.withdraw(10)
    raise AssertionError("withdraw should have raised ValueError")
except ValueError:
    pass

print("Jar tests passed")
```

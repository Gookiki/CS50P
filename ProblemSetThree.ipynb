```python
def checks():
    x, y = input("Fraction: ").split("/")

    if x.isdigit() and y.isdigit() and int(y) >= int(x) and int(y) != 0:
        result = round(int(x) / int(y) * 100)
        if result <= 1:
            print("E")
        elif result >= 99:
            print("F")
        else:
            print(round(int(x) / int(y) * 100), end="%")
    else:
        checks()

checks()
```

```python
menu = {
    "baja taco": 4.25,
    "burrito": 7.50,
    "bowl": 8.50,
    "nachos": 11.00,
    "quesadilla": 8.50,
    "super burrito": 8.50,
    "super quesadilla": 9.50,
    "taco": 3.00,
    "tortilla salad": 8.00
}

total = 0
while True:    
    item = input("Item (q to quit): ").lower()
    if item == "q":
        break
    try:
        total += menu[item]
        print("$" + f"{total:.2f}")
    except KeyError:
        print("Item not found")
```

```python
groceries = {}

while True:
    groceryitem = input("Item (q to quit): ").lower()
    if groceryitem == "q":
        break  
    try:
        groceries[groceryitem] += 1
    except KeyError:
        groceries[groceryitem] = 1

for item in sorted(groceries):
    print(groceries[item], item.upper())
```

```python
months = { "January": 1,
    "February": 2,
    "March": 3,
    "April": 4,
    "May": 5,
    "June": 6,
    "July": 7,
    "August": 8,
    "September": 9,
    "October": 10,
    "November": 11,
    "December": 12
}

def formatchecker(date):
    if date.__contains__("/"):
        try:
            month, day, year = date.split("/")
            if (int(day) > 31 or int(day) < 1) or (int(month) > 12 or int(month) < 1):
                raise ValueError
            else:
                return (f"{year}-{month}-{day}")
        except:
            raise ValueError
    elif date.__contains__(","):
        try:
            monthday, year = date.split(", ")
            monthname, day = monthday.split(" ")
            month = months[monthname]
            if (int(day) > 31 or int(day) < 1) or (int(month) > 12 or int(month) < 1):
                raise ValueError
            else:
                return (f"{year}-{month}-{day}")
        except:
            raise ValueError
    
    else:
        raise ValueError

while True:
    date = input("Date: ")
    if date.lower() == "q":
        break
    try:
       print(formatchecker(date))
    except ValueError as e:
        print("Error")
        
        
    
```

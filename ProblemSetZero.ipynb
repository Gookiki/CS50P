```python
def main():
    price = priceToFloat(input("What was the price? "))
    tip = tipToFloat(input("What % tip would you like to give? "))
    print(price * tip * 0.01)

def priceToFloat(price):
    floatprice = float(price)
    return floatprice

def tipToFloat(tip):
    return float(tip)

main()
    
```
